```python
#Prints the menu for the day of the week, listing the date, cookies, and prices
#Inserted place holder menus for testing purposes
tuesday_menu = {"Nada": {"price": 0, "stock": 0}}
wednesday_menu = {"Nada": {"price": 0, "stock": 0}}
thursday_menu = {"Nada": {"price": 0, "stock": 0}}
friday_menu = {"Nada": {"price": 0, "stock": 0}}
saturday_menu = {"Nada": {"price": 0, "stock": 0}}
sunday_menu = {"Nada": {"price": 0, "stock": 0}}

def print_menu():
  #For testing with the monday menu
  #day_of_the_week = 1
  day_of_the_week = datetime.date.isoweekday(datetime.date.today())
  menus = {
    1: monday_menu,
    2: tuesday_menu,
    3: wednesday_menu,
    4: thursday_menu,
    5: friday_menu,
    6: saturday_menu,
    7: sunday_menu
  }
  menu = menus[day_of_the_week]
  print(f'{datetime.date.today()}')
  for name, details in menu.items():
    print(f'{name}: ${details["price"]}')

print_menu()
```

## Interfacing

```python
def customer_order(menu):
  item = input("What would you like to order? ")
  while item not in menu:
    print("Item is not on the menu today.")
    item = input("What would you like to order? ")
  quantity = int(input("How many would you like? "))
  if menu[item]["stock"] - quantity < 0:
    print("Sorry, there is not enough of that item. We have", menu[item]["stock"], "left.")
    change_quantity = input("Would you like to order a different amount? Enter Y/N: ")
    if change_quantity == "Y":
      quantity = int(input("How many would you like? "))
    elif change_quantity == "N":
      change_cookie = input("Would you like to order a different cookie? Enter Y/N: ")
      if change_cookie == "Y":
        item = input("What would you like to order? ")
      elif change_cookie == "N":
        print("Thanks for coming! Goodbye.")
        return

  menu[item]["stock"] = menu[item]["stock"] - quantity
  order_total = menu[item]["price"] * quantity
  print("Your total is $", order_total, "Thank You for your order!")

customer_order(monday_menu)
```
